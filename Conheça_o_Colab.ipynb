Exercício 1: Gere um gráfico de dispersão baseado em array de números aleatórios entre 0 e 100

```
import numpy as np
import matplotlib.pyplot as plt

# Gera 50 números inteiros aleatórios entre 0 e 100 para o eixo X e Y
x = np.random.randint(0, 101, 50)
y = np.random.randint(0, 101, 50)

# Cria o gráfico de dispersão (scatter plot)
plt.scatter(x, y, color='purple', alpha=0.7, edgecolors='black')

# Adiciona título e rótulos nos eixos
plt.title('Gráfico de Dispersão - Números Aleatórios (0 a 100)')
plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')

# Mostra o gráfico na tela
plt.show()
```

Exercício 2: Escolha um dataset de série temporal do repositório e plote um gráfico de linha ao longo do tempo com a coluna y

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carrega o dataset winequality-white.csv
file_path = '/content/winequality-white.csv'
df_white = pd.read_csv(file_path, sep=';') # O separador é ponto e vírgula

print("Primeiras 5 linhas do DataFrame winequality-white.csv:")
display(df_white.head())
```

```
# Ordena o DataFrame por 'citric acid' para um gráfico de linha mais coerente
df_white_sorted = df_white.sort_values(by='citric acid')

plt.figure(figsize=(12, 7))
sns.lineplot(x='citric acid', y='quality', data=df_white_sorted, marker='o', alpha=0.7)
plt.title('Relação entre Ácido Cítrico e Qualidade do Vinho Branco (Gráfico de Linha)')
plt.xlabel('Ácido Cítrico')
plt.ylabel('Qualidade')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
